```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
```

```python
df = pd.read_csv('/content/sample_data/imdb_top_1000.csv')
df.head()
```

```python
top10 = df.sort_values('IMDB_Rating', ascending=False).head(10)
plt.barh(top10['Series_Title'], top10['IMDB_Rating'], color='skyblue')
plt.xlabel('IMDb Rating')
plt.title('Top 10 IMDb Rated Movies')
plt.gca().invert_yaxis()
plt.show()
```

```python
plt.hist(df['IMDB_Rating'], bins=20, color='orange', edgecolor='black')
plt.title('Distribution of IMDb Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Movies')
plt.grid(True)
plt.show()
```

```python
from collections import Counter

all_genres = df['Genre'].str.split(', ').explode()
top_genres = Counter(all_genres).most_common(5)
labels, sizes = zip(*top_genres)

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.axis('equal')
plt.title('Top 5 Movie Genres')
plt.show()
```

```python
df['Gross'] = pd.to_numeric(df['Gross'].astype(str).str.replace(',', '').str.replace('$', ''), errors='coerce')

plt.scatter(df['IMDB_Rating'], df['Gross'], color='red', alpha=1)
plt.xlabel('IMDb Rating')
plt.ylabel('Gross Earnings (USD)')
plt.title('IMDb Rating vs Gross')
plt.grid(True)
plt.show()
```
